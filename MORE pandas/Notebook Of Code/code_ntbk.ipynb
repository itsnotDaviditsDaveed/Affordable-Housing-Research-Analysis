```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go


DARF = pd.read_csv("Affordable_Housing_Production_by_Building_2023.csv")
print(DARF.columns)
```

```python
ferg_one = px.scatter_map(DARF, lat="Latitude", lon="Longitude", hover_data=["Building ID", "Project ID", "Total Units"], color="Borough", title="All Projects", zoom=8.56)

ferg_one.show()
```

```python
QUEENS_APT = DARF.query("`Borough` == 'Queens'")
qerg = px.scatter_map(QUEENS_APT, lat="Latitude", lon="Longitude", hover_data=["Building ID", "Project ID", "Total Units"], zoom = 10, color = "Street", title="Apartments In Queens")
qerg.update_layout(hoverlabel=dict(
        bgcolor="white",
        font_size=10,
        font_family="Nunito",
        font_color="black",
        bordercolor="white"
    ))
```
